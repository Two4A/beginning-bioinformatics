####**TASK P18.2** <font color= orange>**Complete Rosalind Problem #5  Dictionaries**</font>

````python
with open("rosalind_ini6.txt","r") as data:
  line_content = data.read()

word_counts_manual = {}

# Split the content into words. By default, split() handles multiple spaces and newlines.
words = line_content.split()

for word in words:
  # Check if the word is already in the dictionary
  if word in word_counts_manual:
    # If it is, increment its count
    word_counts_manual[word] += 1
  else:
    # If not, add it with a count of 1
    word_counts_manual[word] = 1

# Print the word counts
for word, count in word_counts_manual.items():
    print(f"{word} {count}")
````

#<font color=blue>**P19.**</font> ```count()``` – counting things made easy

####**TASK P19.1** <font color= orange>**Complete Rosalind Problem #6  Counting DNA Nucleotides**</font>

````python
with open("rosalind_dna.txt","r") as data:
  line_content = data.read().strip() # Read content and remove leading/trailing whitespace

letter_counts_manual = {}

# Convert the string directly into a list of individual characters
letters = list(line_content)

# Get a list of unique characters
unique_letters = list(set(letters))

# Use list.count() to count occurrences of each unique character
for letter in unique_letters:
  # Use the individual 'letter' as the dictionary key and as the item to count
  letter_counts_manual[letter] = letters.count(letter)

# Print the character counts
print(letter_counts_manual["A"],letter_counts_manual["C"],letter_counts_manual["G"],letter_counts_manual["T"])
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
with open("practice_5.txt","r") as data:
  line_content = data.read()

word_counts_manual = {}

# Split the content into words. By default, split() handles multiple spaces and newlines.
words = line_content.split()

# Get a list of unique words
unique_words = list(set(words))

# Use list.count() to count occurrences of each unique word
for word in unique_words:
  word_counts_manual[word] = words.count(word)

# Print the word counts
for word, count in word_counts_manual.items():
    print(f"{word} {count}")
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

####**TASK P23.1** Modify the script from TASK P22.1 so that it prints the GC content for each sequence in the file

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
from Bio.SeqIO import parse
from Bio.SeqUtils import gc_fraction

max_gc_content = -1.0 # Initialize with a value lower than any possible GC content.
highest_gc_id = ""

for seq_record in parse("rosalind_gc.txt", "fasta"):
    current_gc = gc_fraction(seq_record.seq) # gc_fraction needs the sequence object, not the full record.

    # Checks to see if the current DNA string has a greater GC content than the greatest stored DNA string GC content.
    if current_gc > max_gc_content:
        max_gc_content = current_gc
        highest_gc_id = seq_record.id

print(highest_gc_id)
print(max_gc_content*100) # Multiplied by 100 to give a percentage of GC content.
````
